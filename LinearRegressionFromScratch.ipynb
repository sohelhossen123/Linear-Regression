## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
```

## 2. Import Dataset

```python
data = pd.read_table('P176.txt')
```

## 3. Calculate number of observations

```python
n=data.shape[0]
print(n)
```

## 4. Separate the Predictor and Response variable
In linear regression, predictor variables are the inputs used to forecast an outcome, and response variables are the continuous output values you want to predict

```python
x=data['X']
y=data['Y']
```

```python
x.shape
```

## 5. Calculate slope Beta_1

```python
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x*y)
sum_xx = np.sum(x**2)

slope_num = (n*sum_xy) - (sum_x*sum_y)
slope_denom = (n*sum_xx) - (sum_x**2)

beta_1 = slope_num/slope_denom
print(beta_1)
```

## 6. Calculate intercept Beta_0

```python
mean_x = np.mean(x)
mean_y = np.mean(y)

beta_0 = mean_y - (beta_1 * mean_x)
print(beta_0)
```

### Linear Regression equation: y = 14.448 + 0.10536x

## 7. Plotting the Graph

```python
plt.scatter(x,y) #dataset
x1 = np.linspace(0,1800,500)
y1 = beta_0 + beta_1*x1
plt.plot(x1,y1,'-')
plt.show()
```
