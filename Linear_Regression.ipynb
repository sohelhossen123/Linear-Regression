## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

## 2. Import Dataset

```python
data=pd.read_table('P176.txt')
```

```python
data.head()
```

```python
data.shape
```

## 3. X and Y

```python
X=data['X']
Y=data['Y']
```

## 4. Reshape X to 2D array

```python
X.shape
```

```python
X=np.array(X).reshape(-1,1)
X.shape
```

```python
X
```

```python
Y.shape
```

## 5. Initialize Linear Regression Model

```python
lin_reg = LinearRegression()
```

## 6. Fit the Linear Regression Model

```python
lin_reg.fit(X, Y)
```

## 7. Print Coefficients and Intercept

```python
print(f"Coefficient: {lin_reg.coef_} ")
print(f"Intercept: {lin_reg.intercept_}")
```

## 8. Make prediction

```python
Y_pred=lin_reg.predict(X)
Y_pred
```

## 9. Predict for new Data

```python
lin_reg.predict([[250],[1200]])
```

## 10. Evaluate Performance

```python
print(f"Mean squared error: {mean_squared_error(Y,Y_pred)}")
```

```python
print(f"R squared score: {r2_score(Y,Y_pred)}")
```

## 11. Plot Graph

```python
plt.scatter(X, Y, color='blue', label='Test Data')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Regression Line')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression Model Fit')

plt.show()
```
